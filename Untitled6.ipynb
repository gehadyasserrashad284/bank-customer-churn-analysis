#5. Analysis

```python
df['churn'].value_counts(normalize=True) * 100
```

```python
df.groupby('country')['churn'].mean() * 100
```

```python
df.groupby('gender')['churn'].mean() * 100
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[17, 30, 40, 50, 60, 100],
    labels=['18-30', '31-40', '41-50', '51-60', '61+']
)

df.groupby('age_group', observed=True)['churn'].mean() * 100
```

```python
df.groupby('products_number')['churn'].mean() * 100
```

```python
df['products_number'].value_counts().sort_index()
```

```python
df.groupby('active_member')['churn'].mean() * 100
```

```python
df.groupby('credit_card')['churn'].mean() * 100
```

```python
df['balance_group'] = pd.cut(
    df['balance'],
    bins=[-1, 50000, 100000, 150000, 200000, float('inf')],
    labels=['0-50K', '50K-100K', '100K-150K', '150K-200K', '200K+']
)

df.groupby('balance_group', observed=True)['churn'].mean() * 100
```

```python
df['balance_group'].value_counts().sort_index()
```

```python
df['credit_score_group'] = pd.cut(
    df['credit_score'],
    bins=[349, 500, 600, 700, 800, 850],
    labels=['350-500', '501-600', '601-700', '701-800', '801-850']
)

df.groupby('credit_score_group', observed=True)['churn'].mean() * 100
```

```python
df.groupby('tenure')['churn'].mean() * 100
```

#6. Visualization

```python
churn_counts = df['churn'].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(['Stayed', 'Churned'], churn_counts.values)

plt.title('Customer Churn Distribution')
plt.xlabel('Customer Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python
churn_by_country = df.groupby('country')['churn'].mean() * 100

plt.figure(figsize=(7, 4))
plt.bar(churn_by_country.index, churn_by_country.values)

plt.title('Churn Rate by Country')
plt.xlabel('Country')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
churn_by_activity = df.groupby('active_member')['churn'].mean() * 100

plt.figure(figsize=(6, 4))
plt.bar(['Inactive', 'Active'], churn_by_activity.values)

plt.title('Churn Rate by Active Membership')
plt.xlabel('Customer Activity')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
churn_by_gender = df.groupby('gender')['churn'].mean() * 100

plt.figure(figsize=(7, 4))
plt.barh(churn_by_gender.index, churn_by_gender.values)

plt.title('Churn Rate by Gender')
plt.xlabel('Churn Rate (%)')
plt.ylabel('Gender')

plt.show()
```

```python
churn_by_age = df.groupby('age_group', observed=True)['churn'].mean() * 100

plt.figure(figsize=(8, 4))
plt.plot(churn_by_age.index, churn_by_age.values, marker='o')

plt.title('Churn Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
churn_by_products = df.groupby('products_number')['churn'].mean() * 100

plt.figure(figsize=(7, 4))
plt.bar(churn_by_products.index.astype(str), churn_by_products.values)

plt.title('Churn Rate by Number of Products')
plt.xlabel('Number of Products')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
churn_by_balance = df.groupby('balance_group', observed=True)['churn'].mean() * 100

plt.figure(figsize=(8, 4))
plt.bar(churn_by_balance.index, churn_by_balance.values)

plt.title('Churn Rate by Balance Group')
plt.xlabel('Balance Group')
plt.ylabel('Churn Rate (%)')

plt.show()
```
